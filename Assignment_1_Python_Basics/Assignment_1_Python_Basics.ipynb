### Q3. Strings
Accept a string and compute its length, count vowels and consonants, reverse the string, check for palindrome, and extract substrings.

```python
# Navansh Mandowara - 25BEC066
# Q3: String Operations

text = input("Enter a string: ")

print("Length of string:", len(text))

vowels = "aeiouAEIOU"
v_count = sum(1 for ch in text if ch in vowels)
c_count = sum(1 for ch in text if ch.isalpha() and ch not in vowels)

print("Vowels count:    ", v_count)
print("Consonants count:", c_count)

# Reverse string
reversed_text = text[::-1]
print("Reversed string: ", reversed_text)

# Palindrome check
is_palindrome = text.lower() == reversed_text.lower()
print("Is palindrome?   ", is_palindrome)

# Slicing example
print("First 5 characters:", text[:5])
```

### Q5. Conditional Statements & Number Analysis
Implement functions to analyze an integer: sign, parity, factorial calculation, prime check, and Armstrong number verification.

```python
# Navansh Mandowara - 25BEC066
# Q5: Number Analysis & Logic

def is_prime(n):
    if n <= 1:
        return False
    if n <= 3:
        return True
    if n % 2 == 0 or n % 3 == 0:
        return False
    i = 5
    while i * i <= n:
        if n % i == 0 or n % (i + 2) == 0:
            return False
        i += 6
    return True

def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    return sum(int(d) ** power for d in digits) == n

def factorial(n):
    res = 1
    for k in range(2, n + 1):
        res *= k
    return res

val = int(input("Enter an integer to test: "))

# Sign & parity
sign = "Positive" if val > 0 else ("Negative" if val < 0 else "Zero")
parity = "Even" if val % 2 == 0 else "Odd"
print(f"{val} is {sign} and {parity}")

if val >= 0:
    print(f"Factorial of {val} = {factorial(val)}")

print(f"Is {val} prime? {is_prime(val)}")
print(f"Is {val} Armstrong? {is_armstrong(val)}")
```

### Q6. Word Count & String Analysis
Analyze a text sentence to count total words, determine word frequencies, and compute character occurrence distribution.

```python
# Navansh Mandowara - 25BEC066
# Q6: Word and Character Frequency Analysis

sentence = input("Enter a sentence: ")

words = sentence.split()
print("Total words count:", len(words))

# Word frequencies
word_freq = {}
for w in words:
    clean_w = w.strip(".,!?;:").lower()
    word_freq[clean_w] = word_freq.get(clean_w, 0) + 1

print("\nWord Frequencies:")
for w, cnt in word_freq.items():
    print(f"  {w:<12}: {cnt}")

# Character frequencies (excluding whitespace)
char_freq = {}
for ch in sentence:
    if not ch.isspace():
        char_freq[ch] = char_freq.get(ch, 0) + 1

print("\nTop Character Frequencies:")
sorted_chars = sorted(char_freq.items(), key=lambda x: x[1], reverse=True)
for ch, cnt in sorted_chars[:6]:
    print(f"  '{ch}': {cnt}")
```

### Q8. Exception Handling
Demonstrate robust error handling using try-except-else-finally blocks for ZeroDivisionError, ValueError, and IndexError.

```python
# Navansh Mandowara - 25BEC066
# Q8: Exception Handling Demonstration

sample_list = [10, 20, 30, 40]

def test_safe_division(a_str, b_str):
    try:
        a = float(a_str)
        b = float(b_str)
        result = a / b
    except ValueError:
        print(f"Error: Non-numeric input provided ('{a_str}', '{b_str}').")
    except ZeroDivisionError:
        print("Error: Division by zero is mathematically undefined.")
    else:
        print(f"Success: {a} / {b} = {result:.3f}")
    finally:
        print("--- Execution block finalized ---")

print("Test Case 1 (Valid):")
test_safe_division("45", "9")

print("\nTest Case 2 (Zero Division):")
test_safe_division("20", "0")

print("\nTest Case 3 (Invalid Format):")
test_safe_division("100", "abc")

print("\nTest Case 4 (Index Bounds):")
try:
    idx = 10
    val = sample_list[idx]
except IndexError as e:
    print(f"Caught IndexError: Cannot access index {idx} on list of size {len(sample_list)}")
```

### Q9. List Comprehension & Matrices
Use list comprehensions for numerical transformations and implement a 3x3 matrix transposition using nested comprehensions.

```python
# Navansh Mandowara - 25BEC066
# Q9: List Comprehensions & 2D Matrix Operations

# 1. Squares of 1 to 15
squares = [x ** 2 for x in range(1, 16)]
print("Squares 1 to 15:", squares)

# 2. Filtering numbers divisible by 3 or 5 up to 30
multiples = [x for x in range(1, 31) if x % 3 == 0 or x % 5 == 0]
print("Multiples of 3 or 5 (1-30):", multiples)

# 3. 2D Matrix creation and Transposition
matrix_3x3 = [
    [2, 4, 6],
    [8, 10, 12],
    [14, 16, 18]
]

print("\nOriginal 3x3 Matrix:")
for row in matrix_3x3:
    print(" ", row)

# Transpose using nested list comprehension
transpose_3x3 = [[row[col_idx] for row in matrix_3x3] for col_idx in range(len(matrix_3x3[0]))]

print("\nTransposed Matrix:")
for row in transpose_3x3:
    print(" ", row)
```

### Q10. Student Management System
Implement a menu-driven or object-oriented student directory supporting insertion, searching, record updates, and deletion.

```python
# Navansh Mandowara - 25BEC066
# Q10: Student Record Directory System

class StudentDirectory:
    def __init__(self):
        self.records = {}

    def add_student(self, roll_no, name, dept, cgpa):
        if roll_no in self.records:
            print(f"Warning: Roll {roll_no} already exists.")
            return False
        self.records[roll_no] = {
            "name": name,
            "department": dept,
            "cgpa": float(cgpa)
        }
        print(f"Enrolled {name} (Roll: {roll_no}) successfully.")
        return True

    def find_student(self, roll_no):
        return self.records.get(roll_no, None)

    def update_cgpa(self, roll_no, new_cgpa):
        if roll_no in self.records:
            self.records[roll_no]["cgpa"] = float(new_cgpa)
            print(f"Updated CGPA for {self.records[roll_no]['name']} to {new_cgpa}")
        else:
            print(f"Roll {roll_no} not found.")

    def delete_student(self, roll_no):
        if roll_no in self.records:
            deleted = self.records.pop(roll_no)
            print(f"Removed student: {deleted['name']} ({roll_no})")
        else:
            print(f"Roll {roll_no} not found.")

    def display_all(self):
        print(f"\n{'Roll No':<10} | {'Name':<15} | {'Dept':<8} | {'CGPA'}")
        print("-" * 45)
        for r_no, info in self.records.items():
            print(f"{r_no:<10} | {info['name']:<15} | {info['department']:<8} | {info['cgpa']:.2f}")

# Demonstration
portal = StudentDirectory()
portal.add_student("25BEC066", "Navansh M", "ECE", 8.85)
portal.add_student("25BEC010", "Aarav Gupta", "ECE", 8.40)
portal.add_student("25BCS025", "Diya Sharma", "CSE", 9.15)

portal.display_all()

portal.update_cgpa("25BEC010", 8.65)
portal.delete_student("25BCS025")

portal.display_all()
```
